```python
import os
cwd = os.getcwd()
print(cwd)
```

```python
import argparse
def parse_args():
    parser = argparse.ArgumentParser(description='Brisque')
    parser.add_argument('--main_path', default=f"{cwd}/out/32_32Falsellama2_alpaca_output/llama3.1_output_alphca_chat7b_act12_rank32_16_0.0001_0.0/checkpoint-401", help='path to input image file')
    parser.add_argument('--weight_path', default=f'{cwd}/out/32_32Falsellama2_alpaca_output/llama3.1_output_alphca_chat7b_act12_rank32_16_0.0001_0.0lora_ABC.pt', help='path to input image file')
    args = parser.parse_known_args()[0] #parses arguments; each becomes an attribute of args (learning rate = args.lr)
    return vars(args)
args = parse_args()
```

```python
print(args)
```

```python
mainpath = args.get('main_path')
a = load_file(mainpath+'/adapter_model.safetensors',device='cpu')
print("Load a finished")
```

```python
weight_list = torch.load(args.get('weight_path'), map_location=torch.device('cpu'))
print("Load weight list finished")
```

```python
print(weight_list.keys())
```

```python
for idx in range(32):
    a['base_model.model.model.layers.'+str(idx)+'.self_attn.q_proj.lora_B.weight'] = weight_list['q_proj_'+str(idx)+'lora_C'] @ a['base_model.model.model.layers.'+str(idx)+'.self_attn.q_proj.lora_B.weight']
    a['base_model.model.model.layers.'+str(idx)+'.self_attn.v_proj.lora_B.weight'] = weight_list['v_proj_'+str(idx)+'lora_C'] @ a['base_model.model.model.layers.'+str(idx)+'.self_attn.v_proj.lora_B.weight']
print("update finished")
```

```python
save_file(a, mainpath+'/adapter_model.safetensors')
```
